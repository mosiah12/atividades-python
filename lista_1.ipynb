```python
# Par ou ímpar
while True:
  try:
    number = int(input('Digite um número: '))
    break
  except Exception as e:
    print('Erro: ', e)

if number % 2 == 0:
  print('Número par.')
else:
  print('Número ímpar.')
```

```python
# Situação do aluno
while True:
  try:
    score = float(input('Digite a primeira nota: '))
    score2 = float(input('Digite a segunda nota: '))
    break
  except Exception as e:
    print('Erro: ', e)

mean = (score + score2)/2

if mean > 10:
  raise ValueError('A média não pode ser maior que 10.')

if mean >= 7:
  print('Aprovado!')
elif mean >= 5 and mean < 7:
  print('Recuperação.')
else:
  print('Reprovado.')
```

```python
# Maior entre os três números
while True:
  try:
    number1 = float(input('Digite um número: '))
    number2 = float(input('Digite o segundo número: '))
    number3 = float(input('Digite o terceiro número: '))
    break
  except Exception as e:
    print('Erro: ', e)

def get_highest() -> float:
  highest = number1
  if highest < number2:
    highest = number2
  elif highest < number3:
    highest = number3
  return highest

print('Maior número: ', get_highest())
```

```python
# Contagem crescente
while True:
  try:
    number = int(input('Digite um número: '))
    break
  except Exception as e:
    print('Erro: ', e)

for i in range(1, number+1):
  print(i)
```

```python
# Somando de 1 até N
while True:
  try:
    number = int(input('Digite um número: '))
    break
  except Exception as e:
    print('Erro: ', e)

def sum_until(number: int) -> int:
  n = 0
  for i in range(1, number+1):
    n += i
  return n

sum_n = sum_until(number)
print(f'Soma de 1 até {number}: {sum_n}')
```

```python
# Contando números pares
from typing import List
numbers = []
for _ in range(10):
  numbers.append(int(input()))

def get_evens(number_list: List[int]) -> int:
  evens = 0
  for n in number_list:
    if n > 0 and n % 2 == 0:
      evens += 1
  return evens

evens = get_evens(numbers)

print(f'Foram digitados {evens} números pares.')
```

```python
# Soma apenas dos positivos
from typing import List
numbers = []
for _ in range(8):
  numbers.append(int(input()))

def sum_positives(number_list: List[int]) -> int:
  s = 0
  for n in number_list:
    if n > 0:
      s += n
  return s

s = sum_positives(numbers)

print('Soma dos positivos: ', s)
```

```python
# Maior valor digitado
from typing import List
numbers = []
for _ in range(6):
  numbers.append(int(input()))

def greatest_number(number_list: List[int]) -> int:
  highest = number_list[0]
  for number in number_list:
   if number > highest:
    highest = number
  return highest

greatest = greatest_number(numbers)

print('Maior valor: ', greatest)
```

```python
# Tentativas de senha
password = '1234'
while True:
 attempt = input('Insira a senha: ')
 if attempt != password:
  print('Senha inválida, tente novamente.')
 else:
  print('Acesso permitido!')
  break
```

```python
# Número secreto
secret_number = 7

def check_proximity(number: int) -> str:
 if number > secret_number:
   return 'Tente um número menor.'
 elif number < secret_number:
   return 'Tente um número maior.'
 else:
   return 'Parabéns! Você acertou.'

while True:
  try:
    number = int(input('Digite um número: '))
    msg = check_proximity(number)
    print(msg)
    if number == secret_number:
     break
  except Exception as e:
    print('Erro: ', e)
```

```python
# Analisando uma sequência de números
from typing import List, Tuple
sequence = []
def analyze_sequence(seq: List[int]) -> Tuple[int]:
 neg, pos = 0, 0
 s = sum(seq)
 for number in seq:
  if number > 0:
   pos += 1
  elif number < 0:
   neg += 1
 return len(seq), pos, neg, s

while True:
  try:
    number = int(input('Digite um número: '))

    if number != 0:
     sequence.append(number)
    else:
     quant, pos, neg, s = analyze_sequence(sequence)
     print('Quantidade: ', quant)
     print('Positivos: ', pos)
     print('Negativos: ', neg)
     print('Soma: ', s)
     break
  except Exception as e:
    print('Erro: ', e)
```
