# Lista 2 de algoritmos

```python
# Estacionamento
horas = int(input("Digite o número de horas no estacionamento: "))

if horas <= 2:
  valor_a_pagar = 8
elif 2 < horas <= 5:
  valor_a_pagar = 8 + 3*(horas - 2)
else:
  valor_a_pagar = 17 + 2*(horas - 5)

print("Total a pagar: ", valor_a_pagar)
```

```python
# Máquina de venda de ingresso
valor_pago = 0
valor_a_pagar = 25

while valor_a_pagar != 0:
  valor_inserido = int(input("Digite um valor inteiro para inserir na máquina: "))
  valor_a_pagar -= valor_inserido
  if valor_a_pagar > 0:
    print(f"O ingresso não foi totalmente pago, restando: {valor_a_pagar:.2f} R$.\n")
  elif valor_a_pagar < 0:
    print(f"Ingresso pago, devolvendo troco: {valor_a_pagar*-1:.2f} R$.\n")
    break
  else:
    print("Ingresso pago!\n")
    break
```

```python
# Maior valor e sua primeira posição
valores = []

for _ in range(8):
  valores.append(int(input("Digite um número inteiro: ")))

maior_valor = valores[0]
indice_maior_valor = 0

for valor in valores:
  if valor > maior_valor:
    maior_valor = valor

for i, valor in enumerate(valores):
  if valor == maior_valor:
    indice_maior_valor = i
    break

print("\nMaior valor: ", maior_valor)
print("Posição da primeira aparição do maior valor: ", indice_maior_valor + 1)
```

```python
# Segundo maior valor
maior_valor = None
segundo_maior_valor = None

for i in range(8):
  valor = int(input("Digite um número inteiro: "))

  if i == 0:
    maior_valor = valor
  else:
    if valor > maior_valor:
      segundo_maior_valor = maior_valor
      maior_valor = valor
    elif valor < maior_valor:
      if segundo_maior_valor is None or valor > segundo_maior_valor:
        segundo_maior_valor = valor

if segundo_maior_valor is None:
  print("\nMaior valor: ", maior_valor)
  print("Não há segundo maior valor.")
else:
  print("\nMaior valor: ", maior_valor)
  print("Segundo maior valor: ", segundo_maior_valor)
```

```python
# Repetição consecutiva
existe_repetidos = False
anterior = None

valor = int(input("Digite um número inteiro: "))

for i in range(7):
  anterior = valor
  valor = int(input("Digite um número inteiro: "))
  if anterior == valor:
    existe_repetidos = True

if existe_repetidos:
  print("\nExistem valores repetidos consecutivamente.")
else:
  print("\nNão existem valores repetidos consecutivamente.")
```

```python
# Controle de saldo
saldo_atual = 500

def depositar(valor: int, saldo: int) -> int:
    if valor <= 0:
        print("Falha ao realizar o depósito.\n")
        return saldo
    print("Depósito realizado com sucesso.\n")
    return saldo + valor

def sacar(valor: int, saldo: int) -> int:
    if valor > saldo or valor <= 0:
        print("Falha ao realizar o saque.\n")
        return saldo
    print("Saque realizado com sucesso.\n")
    return saldo - valor

print("1 - Depositar")
print("2 - Saque")
print("3 - Sair\n")

for _ in range(6):
    decisao = int(input(">>> "))

    if decisao == 1:
        valor_deposito = int(input("Digite o valor de depósito: "))
        saldo_atual = depositar(valor_deposito, saldo_atual)

    elif decisao == 2:
        valor_saque = int(input("Digite o valor de saque: "))
        saldo_atual = sacar(valor_saque, saldo_atual)

    elif decisao == 3:
        print("Saindo...\n")
        break

    else:
        print("Digite um comando válido.\n")

print(f"Saldo final: {saldo_atual:.2f} R$")
```

```python
# Caixa eletrônico
numero = int(input("Digite um número inteiro positivo: "))
cedulas = 0

def reduzir(valor: int) -> None:
  global numero, cedulas
  while numero - valor >= 0:
      numero -= valor
      cedulas += 1

valores = [100, 50, 20, 10, 5, 2, 1]

for valor in valores:
  reduzir(valor)

print("Quantidade mínima de cédulas a serem usadas: ", cedulas)
```

```python
# Número primo
numero = int(input("Digite um número inteiro maior do que 1: "))
e_primo = True

if numero < 1:
  raise ValueError("O número fornecido é menor do que 1.")

for i in range(2, numero):
  if numero % i == 0:
    e_primo = False

if e_primo:
  print("O número fornecido é primo.")
else:
  print("O número fornecido não é primo.")
```

```python
# Divisores de um número
numero = int(input("Digite um número inteiro: "))
divisores = 0

for i in range(1, numero + 1):
  if numero % i == 0:
    print(f"Divisor de {numero}: {i}")
    divisores += 1

print("Quantidade de divisores: ", divisores)
```

```python
# Números primos em um intervalo
numero1 = int(input("Digite um número inteiro positivo: "))
numero2 = int(input("Digite outro número inteiro positivo: "))

primos = 0

def e_primo(numero: int) -> bool:
  if numero < 2:
    return False
  for i in range(2, numero):
    if numero % i == 0:
      return False
  return True

if numero1 > numero2:
  numero1, numero2 = numero2, numero1

if numero1 <= 2 <= numero2:
  print("Número primo: 2")
  primos += 1

inicio = max(3, numero1)

if inicio % 2 == 0:
  inicio += 1

for i in range(inicio, numero2 + 1, 2):
  if e_primo(i):
    print("Número primo:", i)
    primos += 1

print(f"Foram encontrados {primos} números primos entre {numero1} e {numero2}.")
```

```python
# Jogo com limite de tentativas
numero_secreto = 37
tentativas_maximas = 5

print(f"Você tem {tentativas_maximas} tentativas para adivinhar o número.")

for tentativa in range(1, tentativas_maximas + 1):
  palpite = int(input(f"\nTentativa {tentativa}/{tentativas_maximas} - Digite seu palpite: "))

  if palpite == numero_secreto:
    print(f"Você acertou o número secreto {numero_secreto} em {tentativa} tentativa(s)!")
    break
  elif palpite < numero_secreto:
    print("O número secreto é maior que o seu palpite.")
  else:
    print("O número secreto é menor que o seu palpite.")

print(f"\nVocê esgotou suas tentativas. O número secreto era {numero_secreto}.")
```

```python

# Desafio final - caixa de atendimento
saldo = 200.00
qtd_entradas = 0
qtd_retiradas_realizadas = 0
qtd_retiradas_recusadas = 0
total_entrou = 0.0
total_saiu = 0.0

print(f"Saldo inicial: {saldo:.2f} R$")
print("1 - Entrada")
print("2 - Retirada")
print("0 - Sair")

while True:
  codigo = int(input("\nInforme o código da operação: "))

  if codigo == 0:
    break

  if codigo == 1:
    valor = float(input("Digite o valor da entrada: R$ "))
    saldo += valor
    total_entrou += valor
    qtd_entradas += 1
    print(f"Entrada realizada. Saldo atual: R$ {saldo:.2f}")

  elif codigo == 2:
    valor = float(input("Digite o valor da retirada: R$ "))

    if valor <= saldo:
      saldo -= valor
      total_saiu += valor
      qtd_retiradas_realizadas += 1
      print(f"Retirada realizada. Saldo atual: R$ {saldo:.2f}")
    else:
      qtd_retiradas_recusadas += 1
      print(f"Retirada recusada. Saldo insuficiente.")

  else:
    print("Código inválido.")

print(f"Saldo final: R$ {saldo:.2f}")
print(f"Quantidade de entradas: {qtd_entradas}")
print(f"Quantidade de retiradas realizadas: {qtd_retiradas_realizadas}")
print(f"Quantidade de retiradas recusadas: {qtd_retiradas_recusadas}")
print(f"Total que entrou: R$ {total_entrou:.2f}")
print(f"Total que saiu: R$ {total_saiu:.2f}")
```
