```python
# Lista de números pares e ímpares
numeros = []

for i in range(1, 11):
  numero = int(input(f"Digite um número inteiro ({i}/10): "))
  numeros.append(numero)

impares, pares = [], []

for numero in numeros:
  if numero >= 0 and numero % 2 == 0:
    pares.append(numero)
  elif numero > 0 and numero % 2 != 0:
    impares.append(numero)
  else:
    continue

print("\nNúmeros digitados:", numeros)
print("Números ímpares:", impares)
print("Números pares:", pares)
```

```python
# Maior e menor da lista
numeros = []

for i in range(1, 9):
  numero = int(input(f"Digite um número inteiro ({i}/8): "))
  numeros.append(numero)

maior = numeros[0]
pos_maior = 0

menor = numeros[0]
pos_menor = 0

for i, numero in enumerate(numeros):
  if numero > maior:
    maior = numero
    pos_maior = i
  if numero < menor:
    menor = numero
    pos_menor = i

print(f"\nMaior número: {maior} (Posição {pos_maior + 1})")
print(f"\nMenor número: {menor} (Posição {pos_menor + 1})")
```

```python
# Nomes sem repetição
nomes = []

print("Digite FIM para sair.\n")
entrada = input("Digite um nome: ")

nomes.append(entrada)

while entrada.lower() != "fim":
  entrada = input("Digite um nome: ")
  if entrada not in nomes and entrada.lower() != "fim":
    nomes.append(entrada)

print("\nNomes diferentes digitados:", nomes)
print("Quantidade:", len(nomes))
```

```python
# Dias da semana em uma tupla
dias = ("Segunda-feira", "Terça-feira", "Quarta-feira", "Quinta-feira", "Sexta-feira", "Sábado", "Domingo")

numero = int(input("Digite um número de 1 a 7: "))

if numero >= 1 and numero <= 7:
  print("Dia correspondente ao número:", dias[numero - 1])
if numero <= 0 or numero > 7:
  print("\nO número digitado não é válido.")
```

```python
numeros = []

for i in range(1, 7):
  numero = int(input(f"Digite um número inteiro ({i}/6): "))
  numeros.append(numero)

numeros = tuple(numeros)

apareceu_5 = False

quant_5 = 0
quant_pares = 0

pos_5 = -1

for i, numero in enumerate(numeros):
  if numero == 5:
    quant_5 += 1
    if apareceu_5 == False:
      pos_5 = i + 1
      apareceu_5 = True
  if numero % 2 == 0 and numero >= 0:
    quant_pares += 1

print("\nQuantidade de vezes que o número cinco apareceu:", quant_5)

if apareceu_5 == True:
  print(f"Primeira ocorrência do número cinco: {pos_5}º posição")

print("\nQuantidade de valores pares:", quant_pares)
```

```python
# Dicionário de aluno
dados = {}

nome = input("Digite o nome do aluno: ")
turma = input("Digite a turma do aluno: ")

nota1 = float(input("Digite a primeira nota do aluno: "))
if nota1 < 0 or nota1 > 10:
  raise ValueError("A nota deve ser maior que 0 e menor ou igual a 10.")

nota2 = float(input("Digite a segunda nota do aluno: "))
if nota2 < 0 or nota2 > 10:
  raise ValueError("A nota deve ser maior que 0 e menor ou igual a 10.")

media = (nota1 + nota2) / 2

dados["Nome"] = nome
dados["Turma"] = turma

dados["Primeira nota"] = nota1
dados["Segunda nota"] = nota2
dados["Média"] = media

print()

for chave, valor in dados.items():
  print(f"{chave}: {valor}")

print()

if media >= 7:
  print("O aluno está aprovado.")
elif media >= 5:
  print("O aluno está de recuperação.")
elif media < 5:
  print("O aluno está reprovado.")
```

```python
# Frequência das palavras

frequencia = {}

for i in range(8):
  palavra = input("Digite uma palavra: ")
  if palavra in frequencia:
    frequencia[palavra] += 1
  else:
    frequencia[palavra] = 1

for palavra, qtd in frequencia.items():
  print(f"{palavra}: {qtd}")
```

```python
# Estoque e atualização

estoque = {
  "caneta": 50,
  "caderno": 20,
  "borracha": 30,
  "lapis": 45
}

produto = input("Digite o nome do produto: ")
quantidade = int(input("Digite a quantidade vendida: "))

if produto not in estoque:
  print("Produto não cadastrado.")
elif estoque[produto] < quantidade:
  print("Estoque insuficiente.")
else:
  estoque[produto] -= quantidade
  print("Venda realizada com sucesso.")

print("Estoque final:", estoque)
```

```python
# Cadastro de alunos

turma = []

for i in range(5):
  nome = input("Digite o nome do aluno: ")
  nota1 = float(input("Digite a nota 1: "))
  nota2 = float(input("Digite a nota 2: "))
  media = (nota1 + nota2) / 2

  aluno = {
    "nome": nome,
    "notas": [nota1, nota2],
    "media": media
  }
  turma.append(aluno)

for aluno in turma:
  print(f"Aluno: {aluno['nome']} | Média: {aluno['media']:.2f}")
```

```python
3
# Catálogo com código

catalogo = []

for i in range(5):
  codigo = int(input("Digite o código do produto: "))
  nome = input("Digite o nome do produto: ")
  preco = float(input("Digite o preço do produto: "))
  produto = (codigo, nome, preco)
  catalogo.append(produto)

codigo_busca = int(input("Digite o código para busca: "))
encontrado = False

for produto in catalogo:
  if produto[0] == codigo_busca:
    print(f"Nome: {produto[1]} | Preço: R$ {produto[2]:.2f}")
    encontrado = True
    break

if not encontrado:
  print("Código não encontrado.")
```

```python
# Sistema de estoque

estoque = []

for i in range(5):
  codigo = int(input("Digite o código: "))
  nome = input("Digite o nome: ")
  quantidade = int(input("Digite a quantidade: "))
  preco = float(input("Digite o preço: "))

  item = {
    "codigo": codigo,
    "nome": nome,
    "quantidade": quantidade,
    "preco": preco
  }
  estoque.append(item)

print("\nItens no estoque:")
for item in estoque:
  print(item)

valor_total_estoque = 0.0
maior_valor_item = -1.0
item_maior_valor = None

for item in estoque:
  valor_item = item["quantidade"] * item["preco"]
  valor_total_estoque += valor_item

  if valor_item > maior_valor_item:
    maior_valor_item = valor_item
    item_maior_valor = item

resumo = (len(estoque), valor_total_estoque)

print(f"\nValor total do estoque: R$ {valor_total_estoque:.2f}")
print(f"Item de maior valor em estoque: {item_maior_valor['nome']} (R$ {maior_valor_item:.2f})")
print(f"Tupla resumo (quantidade itens, valor total): {resumo}")
```
