```python
# Maior sequência de crescimento

nums = []

maior_quant = 0
quant = 0

for i in range(10):
  atual = int(input(f"Digite um número inteiro ({i + 1}/{10}): "))
  nums.append(atual)

  if i == 0:
    quant = 1
  elif atual > nums[i - 1]:
    quant += 1
  else:
    quant = 1

  if quant > maior_quant:
    maior_quant = quant

print(f"Maior número de elementos consecutivos: {maior_quant}")
```

```python
# Pontos no plano

pontos = [(1, 2), (-3, 7), (9, 0),
          (0, 0), (-2, -2), (6, -1)]

def quadrante(ponto: tuple):
  x, y = ponto[0], ponto[1]
  if x == 0 or y == 0:
    return 0
  elif x > 0 and y > 0:
    return 1
  elif x < 0 and y > 0:
    return 2
  elif x < 0 and y < 0:
    return 3
  return 4

def soma_quadrados(ponto: tuple):
  x, y = ponto[0], ponto[1]
  return (x**2) + (y**2)

quadrante1 = 0
quadrante2 = 0
quadrante3 = 0
quadrante4 = 0

maior = pontos[0]

for ponto in pontos:
  if soma_quadrados(ponto) > soma_quadrados(maior):
    maior = ponto

  if quadrante(ponto) == 1:
    quadrante1 += 1
  elif quadrante(ponto) == 2:
    quadrante2 += 1
  elif quadrante(ponto) == 3:
    quadrante3 += 1
  elif quadrante(ponto) == 4:
    quadrante4 += 1

print(f"Ponto com a maior soma de quadrados: {maior}")

print(f"\nQuantidade de pontos no 1º quadrante: {quadrante1}")
print(f"Quantidade de pontos no 2º quadrante: {quadrante2}")
print(f"Quantidade de pontos no 3º quadrante: {quadrante3}")
print(f"Quantidade de pontos no 4º quadrante: {quadrante4}")
```

```python
# Frequência e palavra mais repetida

palavras = []
for i in range(10):
  palavra = input(f"Digite a {i+1}ª palavra: ").lower()
  palavras.append(palavra)

frequencias = {}
for palavra in palavras:
  if palavra in frequencias:
    frequencias[palavra] += 1
  else:
    frequencias[palavra] = 1

print("\nDicionário de Frequências:")
for palavra, freq in frequencias.items():
  print(f"'{palavra}': {freq}")

maior_frequencia = 0
for freq in frequencias.values():
  if freq > maior_frequencia:
    maior_frequencia = freq

palavras_mais_repetidas = []
for palavra, freq in frequencias.items():
  if freq == maior_frequencia:
    palavras_mais_repetidas.append(palavra)

print(f"\nMaior frequência encontrada: {maior_frequencia}")
print(f"Palavras com a maior frequência: {palavras_mais_repetidas}")
```

```python
# Boletim de uma turma

turma = []

for i in range(6):
  print(f"\nCadastro do Aluno {i+1}:")
  nome = input("Digite o nome do aluno: ")
  notas = []
  for j in range(3):
    nota = float(input(f"Digite a {j+1}ª nota: "))
    notas.append(nota)

  soma_notas = 0
  for nota in notas:
    soma_notas += nota
  media_aluno = soma_notas / len(notas)

  aluno = {
      "nome": nome,
      "notas": notas,
      "media": media_aluno
  }
  turma.append(aluno)

print("\n--- Boletim --- ")

total_medias_turma = 0
for aluno in turma:
  print(f"Aluno: {aluno['nome']}, Média: {aluno['media']:.2f}")
  total_medias_turma += aluno['media']

media_geral_turma = total_medias_turma / len(turma)
print(f"\nMédia geral da turma: {media_geral_turma:.2f}")

alunos_acima_media_turma = 0
for aluno in turma:
  if aluno['media'] >= media_geral_turma:
    alunos_acima_media_turma += 1

print(f"Quantidade de alunos com média maior ou igual à média da turma: {alunos_acima_media_turma}")

maior_media = 0
alunos_maior_media = []

for aluno in turma:
  if aluno['media'] > maior_media:
    maior_media = aluno['media']
    alunos_maior_media = [aluno['nome']]
  elif aluno['media'] == maior_media:
    alunos_maior_media.append(aluno['nome'])

print(f"\nAlunos com a maior média ({maior_media:.2f}): {alunos_maior_media}")
```

```python
# Controle de estoque com movimentações

produtos = []

for i in range(5):
  print(f"\nCadastro do Produto {i+1}:")
  codigo = input("Digite o código do produto: ")
  nome = input("Digite o nome do produto: ")
  preco = float(input("Digite o preço do produto: "))
  quantidade = int(input("Digite a quantidade inicial do produto: "))

  produto = {
      "codigo": codigo,
      "nome": nome,
      "preco": preco,
      "quantidade": quantidade
  }
  produtos.append(produto)

print("\n--- Realizando 5 Operações de Movimentação ---")

for i in range(5):
  print(f"\nOperação {i+1}:")
  op_codigo = input("Digite o código do produto para movimentação: ")
  op_tipo = input("Digite o tipo de operação (E para Entrada, S para Saída): ").upper()
  op_quantidade = int(input("Digite a quantidade da movimentação: "))

  produto_encontrado = None
  for produto in produtos:
    if produto["codigo"] == op_codigo:
      produto_encontrado = produto
      break

  if produto_encontrado:
    if op_tipo == 'E':
      produto_encontrado["quantidade"] += op_quantidade
      print(f"Entrada de {op_quantidade} unidades para o produto {produto_encontrado['nome']} realizada com sucesso.")
    elif op_tipo == 'S':
      if produto_encontrado["quantidade"] >= op_quantidade:
        produto_encontrado["quantidade"] -= op_quantidade
        print(f"Saída de {op_quantidade} unidades para o produto {produto_encontrado['nome']} realizada com sucesso.")
      else:
        print(f"Operação recusada: Estoque insuficiente para o produto {produto_encontrado['nome']}.")
    else:
      print("Operação recusada: Tipo de operação inválido. Use 'E' ou 'S'.")
  else:
    print(f"Operação recusada: Produto com código '{op_codigo}' não encontrado.")

print("\n--- Estoque Atualizado ---")

valor_financeiro_total = 0
for produto in produtos:
  print(f"Produto: {produto['nome']} (Código: {produto['codigo']}) - Quantidade: {produto['quantidade']} - Preço: R${produto['preco']:.2f}")
  valor_financeiro_total += produto['quantidade'] * produto['preco']

print(f"\nValor financeiro total do estoque: R${valor_financeiro_total:.2f}")
```

```python
# Ranking de jogadores

jogadores = []

for i in range(5):
  print(f"\nCadastro do Jogador {i+1}:")
  nome = input("Digite o nome do jogador: ")
  pontuacoes = []
  for j in range(3):
    pontuacao = int(input(f"Digite a pontuação da {j+1}ª partida: "))
    pontuacoes.append(pontuacao)

  total_pontos = 0
  for p in pontuacoes:
    total_pontos += p
  media_pontos = total_pontos / len(pontuacoes)

  maior_pontuacao_jogador = pontuacoes[0]
  menor_pontuacao_jogador = pontuacoes[0]
  for p in pontuacoes:
    if p > maior_pontuacao_jogador:
      maior_pontuacao_jogador = p
    if p < menor_pontuacao_jogador:
      menor_pontuacao_jogador = p

  diferenca_pontuacoes = maior_pontuacao_jogador - menor_pontuacao_jogador

  jogador = {
      "nome": nome,
      "pontuacoes": pontuacoes,
      "total": total_pontos,
      "media": media_pontos,
      "diferenca": diferenca_pontuacoes
  }
  jogadores.append(jogador)

print("\n--- Relatório de Jogadores ---")

for jogador in jogadores:
  print(f"\nJogador: {jogador['nome']}")
  print(f"  Pontuações: {jogador['pontuacoes']}")
  print(f"  Total: {jogador['total']}")
  print(f"  Média: {jogador['media']:.2f}")

maior_total = 0
jogadores_maior_total = []

for jogador in jogadores:
  if jogador['total'] > maior_total:
    maior_total = jogador['total']
    jogadores_maior_total = [jogador['nome']]
  elif jogador['total'] == maior_total:
    jogadores_maior_total.append(jogador['nome'])

print(f"\nJogadores com maior total de pontos ({maior_total}): {jogadores_maior_total}")

menor_diferenca = jogadores[0]['diferenca'] if jogadores else 0
jogadores_mais_regulares = []

if jogadores:
  for jogador in jogadores:
    if jogador['diferenca'] < menor_diferenca:
      menor_diferenca = jogador['diferenca']
      jogadores_mais_regulares = [jogador['nome']]
    elif jogador['diferenca'] == menor_diferenca:
      jogadores_mais_regulares.append(jogador['nome'])

print(f"Jogadores mais regulares (menor diferença {menor_diferenca}): {', '.join(jogadores_mais_regulares)}")
```

```python
# Agenda de contatos sem duplicidade

agenda = []

while True:
  print("\n--- Cadastro de Contato ---")
  nome = input("Digite o nome do contato (ou FIM para encerrar o cadastro): ")
  if nome.upper() == 'FIM':
    break

  nome_existente = False
  for contato in agenda:
    if contato['nome'].lower() == nome.lower():
      nome_existente = True
      break

  if nome_existente:
    print(f"Erro: Um contato com o nome '{nome}' já existe. Por favor, digite outro nome.")
  else:
    telefone = input(f"Digite o telefone de {nome}: ")
    novo_contato = {
        "nome": nome,
        "telefone": telefone
    }
    agenda.append(novo_contato)
    print(f"Contato '{nome}' cadastrado com sucesso.")

print("\n--- Consulta de Contatos ---")

while True:
  nome_consulta = input("Digite o nome do contato para consultar (ou FIM para encerrar a consulta): ")
  if nome_consulta.upper() == 'FIM':
    break

  contato_encontrado = False
  for contato in agenda:
    if contato['nome'].lower() == nome_consulta.lower():
      print(f"Nome: {contato['nome']}, Telefone: {contato['telefone']}")
      contato_encontrado = True
      break

  if not contato_encontrado:
    print(f"Contato '{nome_consulta}' não encontrado na agenda.")

print("\nFim da Agenda.")
```

```python
# Pesquisa de votos

votos = {
    1: 0,
    2: 0,
    3: 0,
    4: 0
}

print("\n--- Pesquisa de Votos ---")
print("Vote nas opções 1 a 4. Digite 0 para encerrar.")

total_votos_validos = 0
while True:
  try:
    voto = int(input("Digite seu voto (1-4, ou 0 para encerrar): "))
    if voto == 0:
      break
    elif voto in votos:
      votos[voto] += 1
      total_votos_validos += 1
    else:
      print("Voto inválido. Por favor, digite um número entre 1 e 4.")
  except ValueError:
    print("Entrada inválida. Por favor, digite um número.")

print("\n--- Resultados da Pesquisa ---")

if total_votos_validos == 0:
  print("Nenhum voto válido foi registrado.")
else:
  maior_votacao = 0
  for count in votos.values():
    if count > maior_votacao:
      maior_votacao = count

  opcoes_vencedoras = []
  for opcao, count in votos.items():
    percentual = (count / total_votos_validos) * 100
    print(f"Opção {opcao}: {count} voto(s) ({percentual:.2f}%)")
    if count == maior_votacao and maior_votacao > 0:
      opcoes_vencedoras.append(opcao)

  if opcoes_vencedoras:
    print(f"\nOpção(ões) vencedora(s): {opcoes_vencedoras} com {maior_votacao} voto(s).")
  else:
    print("\nNão houve opções vencedoras (nenhum voto ou todos empatados com zero).")
```

```python
# Histórico de movimentações

estoque = {
    "PROD001": {"nome": "Produto A", "quantidade": 50},
    "PROD002": {"nome": "Produto B", "quantidade": 30},
    "PROD003": {"nome": "Produto C", "quantidade": 20}
}

historico_movimentacoes = []

movimentacoes_programadas = [
    ("PROD001", "E", 10),
    ("PROD002", "S", 15),
    ("PROD003", "S", 5),
    ("PROD001", "S", 70),
    ("PROD004", "E", 100),
    ("PROD002", "E", 20),
    ("PROD003", "S", 20),
    ("PROD001", "S", 5)
]

print("\n--- Histórico de Movimentações --- ")
for i, (codigo, tipo, quantidade) in enumerate(movimentacoes_programadas):
  print(f"\nProcessando movimentação {i+1}: ({codigo}, {tipo}, {quantidade})")
  status = "Sucesso"
  if codigo in estoque:
    if tipo == 'E':
      estoque[codigo]["quantidade"] += quantidade
    elif tipo == 'S':
      if estoque[codigo]["quantidade"] >= quantidade:
        estoque[codigo]["quantidade"] -= quantidade
      else:
        status = "Recusada: Estoque insuficiente"
    else:
      status = "Recusada: Tipo de operação inválido"
  else:
    status = "Recusada: Código de produto não encontrado"
  historico_movimentacoes.append((codigo, tipo, quantidade, status))
  print(f"Status: {status}")

print("\n--- Estoque Final ---")
for codigo, dados in estoque.items():
  print(f"Código: {codigo}, Nome: {dados['nome']}, Quantidade: {dados['quantidade']}")

print("\n--- Histórico Completo de Movimentações ---")
for mov in historico_movimentacoes:
  print(f"Movimentação: Código={mov[0]}, Tipo={mov[1]}, Quantidade={mov[2]}, Status={mov[3]}")
```

```python
# Análise de vendas por vendedor

vendedores = []

for i in range(5):
  print(f"\nCadastro do Vendedor {i + 1}:")
  nome_vendedor = input("Digite o nome do vendedor: ")
  vendas_vendedor = []
  for j in range(4):
    valor_venda = float(input(f"Digite o valor da {j + 1}ª venda: "))
    vendas_vendedor.append(valor_venda)

  vendedor = {
      "nome": nome_vendedor,
      "vendas": vendas_vendedor
  }
  vendedores.append(vendedor)

print("\n--- Análise de Vendas ---")

totais_vendedores = []
total_geral_vendas = 0

for vendedor in vendedores:
  total_vendas_vendedor = 0
  for venda in vendedor["vendas"]:
    total_vendas_vendedor += venda
  vendedor["total_vendido"] = total_vendas_vendedor
  totais_vendedores.append(total_vendas_vendedor)
  total_geral_vendas += total_vendas_vendedor
  print(f"Vendedor: {vendedor['nome']}, Total Vendido: R${vendedor['total_vendido']:.2f}")

print(f"\nTotal geral de vendas da equipe: R${total_geral_vendas:.2f}")

media_vendas_equipe = 0
if vendedores:
  media_vendas_equipe = total_geral_vendas / len(vendedores)
print(f"Média de vendas por vendedor na equipe: R${media_vendas_equipe:.2f}")

maior_total_vendas = 0
vendedores_maior_total = []

for vendedor in vendedores:
  if vendedor["total_vendido"] > maior_total_vendas:
    maior_total_vendas = vendedor["total_vendido"]
    vendedores_maior_total = [vendedor["nome"]]
  elif vendedor["total_vendido"] == maior_total_vendas:
    vendedores_maior_total.append(vendedor["nome"])

if vendedores_maior_total:
  print(f"\nVendedor(es) com maior total de vendas (R${maior_total_vendas:.2f}): {', '.join(vendedores_maior_total)}")
else:
  print("\nNão há vendedores para analisar.")

vendedores_acima_media = 0
for vendedor in vendedores:
  if vendedor["total_vendido"] > media_vendas_equipe:
    vendedores_acima_media += 1

print(f"Quantidade de vendedores acima da média de vendas da equipe: {vendedores_acima_media}")
```

```python
# Catálogo e pedidos

produtos = [
    {"codigo": "P001", "nome": "TV", "preco": 2500.00, "estoque": 10},
    {"codigo": "P002", "nome": "Smartphone", "preco": 1200.00, "estoque": 25},
    {"codigo": "P003", "nome": "Notebook", "preco": 3500.00, "estoque": 7},
    {"codigo": "P004", "nome": "Fones", "preco": 300.00, "estoque": 50},
    {"codigo": "P005", "nome": "Smartwatch", "preco": 800.00, "estoque": 15}
]

pedidos = [
    ("P001", 2),
    ("P002", 10),
    ("P006", 5),
    ("P003", 2),
    ("P001", 12),
    ("P004", 20)
]

valor_total_pedidos_atendidos = 0

print("\n--- Processamento de Pedidos ---")
for i, (codigo_pedido, quantidade_pedido) in enumerate(pedidos):
  print(f"\nPedido {i + 1}: Código={codigo_pedido}, Quantidade={quantidade_pedido}")
  produto_encontrado = None
  for produto in produtos:
    if produto["codigo"] == codigo_pedido:
      produto_encontrado = produto
      break

  if produto_encontrado:
    if produto_encontrado["estoque"] >= quantidade_pedido:
      produto_encontrado["estoque"] -= quantidade_pedido
      valor_pedido = produto_encontrado["preco"] * quantidade_pedido
      valor_total_pedidos_atendidos += valor_pedido
      print(f"  Status: ATENDIDO. Valor do pedido: R${valor_pedido:.2f}")
    else:
      print(f"  Status: RECUSADO. Estoque insuficiente para {produto_encontrado['nome']}. Estoque atual: {produto_encontrado['estoque']}")
  else:
    print(f"  Status: RECUSADO. Produto com código '{codigo_pedido}' não encontrado no catálogo.")

print("\n--- Catálogo e Estoque Atualizado ---")
for produto in produtos:
  print(f"Produto: {produto['nome']} (Código: {produto['codigo']}) - Preço: R${produto['preco']:.2f}, Estoque: {produto['estoque']}")

print(f"\nValor total dos pedidos atendidos: R${valor_total_pedidos_atendidos:.2f}")
```

```python
# Sistema de vendas e relatório final

produtos = [
    {"codigo": "P001", "nome": "Caneta", "preco": 2.50, "quantidade": 100},
    {"codigo": "P002", "nome": "Caderno", "preco": 15.00, "quantidade": 50},
    {"codigo": "P003", "nome": "Mochila", "preco": 80.00, "quantidade": 20},
    {"codigo": "P004", "nome": "Lápis", "preco": 1.00, "quantidade": 200},
    {"codigo": "P005", "nome": "Borracha", "preco": 0.75, "quantidade": 150}
]

historico_vendas = []

print("\n--- Sistema de Vendas ---")
while True:
  try:
    codigo_venda = input("Digite o código do produto para venda (ou 0 para encerrar): ")
    if codigo_venda == '0':
      break

    quantidade_venda = int(input(f"Digite a quantidade a ser vendida de {codigo_venda}: "))

    produto_encontrado = None
    for produto in produtos:
      if produto["codigo"] == codigo_venda:
        produto_encontrado = produto
        break

    if produto_encontrado:
      if quantidade_venda > 0:
        if produto_encontrado["quantidade"] >= quantidade_venda:
          produto_encontrado["quantidade"] -= quantidade_venda
          valor_total_venda = produto_encontrado["preco"] * quantidade_venda
          historico_vendas.append((codigo_venda, quantidade_venda, valor_total_venda))
          print(f"Venda de {quantidade_venda} unidades de {produto_encontrado['nome']} por R${valor_total_venda:.2f} registrada.")
        else:
          print(f"Estoque insuficiente para {produto_encontrado['nome']}. Disponível: {produto_encontrado['quantidade']}")
      else:
        print("Quantidade de venda deve ser maior que zero.")
    else:
      print(f"Produto com código '{codigo_venda}' não encontrado.")
  except ValueError:
    print("Entrada inválida. Digite um número para a quantidade.")

print("\n--- Relatório Final de Vendas ---")

print("\n--- Estoque Restante ---")
for produto in produtos:
  print(f"Produto: {produto['nome']} (Código: {produto['codigo']}) - Estoque: {produto['quantidade']}")

quantidade_total_vendida = 0
valor_total_todas_vendas = 0
produtos_vendidos_valores = {}

for venda in historico_vendas:
  quantidade_total_vendida += venda[1]
  valor_total_todas_vendas += venda[2]

  codigo = venda[0]
  valor = venda[2]
  if codigo in produtos_vendidos_valores:
    produtos_vendidos_valores[codigo] += valor
  else:
    produtos_vendidos_valores[codigo] = valor

print(f"\nTotal de unidades vendidas: {quantidade_total_vendida}")
print(f"Valor total de todas as vendas: R${valor_total_todas_vendas:.2f}")

maior_valor_vendas_produto = 0
produtos_maior_valor_vendas = []

for codigo, valor in produtos_vendidos_valores.items():
  if valor > maior_valor_vendas_produto:
    maior_valor_vendas_produto = valor
    produtos_maior_valor_vendas = [codigo]
  elif valor == maior_valor_vendas_produto:
    produtos_maior_valor_vendas.append(codigo)

if produtos_maior_valor_vendas:
  nomes_produtos_maior_valor = []
  for codigo in produtos_maior_valor_vendas:
    for produto in produtos:
      if produto['codigo'] == codigo:
        nomes_produtos_maior_valor.append(produto['nome'])
        break
  print(f"\nProduto(s) de maior valor em vendas (R${maior_valor_vendas_produto:.2f}): {', '.join(nomes_produtos_maior_valor)}")
else:
  print("\nNão houve produtos com vendas registradas.")

produtos_sem_venda = []
for produto in produtos:
  vendido = False
  for venda in historico_vendas:
    if venda[0] == produto['codigo']:
      vendido = True
      break
  if not vendido:
    produtos_sem_venda.append(produto['nome'])

if produtos_sem_venda:
  print(f"\nProdutos sem vendas: {', '.join(produtos_sem_venda)}")
else:
  print("\nTodos os produtos tiveram vendas.")

print("\n--- Histórico de Vendas ---")
for venda_registro in historico_vendas:
  print(f"Código: {venda_registro[0]}, Quantidade: {venda_registro[1]}, Valor: R${venda_registro[2]:.2f}")

tupla_resumo = (len(historico_vendas), quantidade_total_vendida, valor_total_todas_vendas)
print(f"\nTupla resumo (quantidade_de_vendas, unidades_vendidas, valor_total): {tupla_resumo}")
```
